**1. Find out the list of most popular and liked genre**

```python
merged_df = pd.merge(netflix_dataset, df_title, left_on='Movie_ID', right_on='movieId')
merged_df
```

```python
movie_ratings = netflix_dataset.groupby('Movie_ID')['Rating'].mean().reset_index()
```

```python
merged_df = pd.merge(movie_ratings, df_title, left_on='Movie_ID', right_on='movieId')
```

```python
merged_df['genres'] = merged_df['genres'].str.split('|')
genre_df = merged_df.explode('genres')
```

```python
popular_genres = genre_df.groupby('genres')['Rating'].mean().sort_values(ascending=False)

print(popular_genres)
```
